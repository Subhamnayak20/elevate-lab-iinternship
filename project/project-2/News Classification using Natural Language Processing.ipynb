```python
import nltk
import pandas as pd
```

```python
fake=pd.read_csv("Fake.csv")
genuine = pd.read_csv("real.csv")
```

```python
genuine
```

```python
fake['genuiness']=0
genuine['genuiness']=1
```

```python
genuine
```

```python
dataset=pd.concat([fake,genuine])
```

```python
dataset
```

## Pre Processing

```python
dataset = dataset.reset_index(drop=True)
```

```python
dataset
```

```python
dataset = dataset.drop(['date','title','subject'],axis=1)
```

```python
dataset
```

```python
from nltk.tokenize import word_tokenize
dataset['text'] = dataset['text'].apply(word_tokenize)
```

```python
dataset
```

```python
from nltk.stem.snowball import SnowballStemmer
```

```python
sb = SnowballStemmer('english')
```

```python
def stem_it(text):
    return [sb.stem(word) for word in text]
```

```python
dataset['text'] = dataset['text'].apply(stem_it)
```

```python
dataset
```

```python
def stopword_revomal(text):
    return [word for word in text if len(word)>>2]
```

```python
dataset['text'] = dataset['text'].apply(stopword_revomal)
```

```python
dataset
```

```python
dataset['text'] = dataset['text'].apply(' '.join)
```

```python
dataset
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(dataset['text'],dataset['genuiness'],test_size=0.25)
```

```python
x_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidfv = TfidfVectorizer(max_df = 0.7)
```

```python
tfidfv_train = tfidfv.fit_transform(x_train)
tfidfv_test = tfidfv.transform(x_test)
```

```python
print(tfidfv_train)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression(max_iter = 1000)
lr.fit(tfidfv_train,y_train)
```

```python
y_pred = lr.predict(tfidfv_test)
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
from sklearn.linear_model import PassiveAggressiveClassifier

pac = PassiveAggressiveClassifier(max_iter = 100)
pac.fit(tfidfv_train,y_train)
```

```python
y_pred_pac = pac.predict(tfidfv_test)
accuracy_score(y_test,y_pred_pac)
```

```python
# Sample news text
sample_news = ["The government announced a new AI policy for schools."]

# Transform using the same TF-IDF vectorizer used in training
sample_vector = tfidfv.transform(sample_news)

# Predict using the trained model
prediction = pac.predict(sample_vector)
label_map = {0: "REAL", 1: "FAKE"}
print("Predicted Label:", label_map[prediction[0]])
```

```python
def test_news(text):
    vector = tfidfv.transform([text])  # Wrap text in a list
    prediction = pac.predict(vector)
    label_map = {0: "REAL", 1: "FAKE"}
    return f"Predicted Label: {label_map[prediction[0]]}"
```

```python
sample1 = "MOSCOW (Reuters) - Vatican Secretary of State."
sample2 = "The government announced a new AI policy for schools."

print("sample1: ",test_news(sample1))
print("sample2: ",test_news(sample2))
```
